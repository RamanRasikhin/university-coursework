```python
s="Ala ma kota."
k=4
def enc(mess, key):
    cipher= "" 
    for i in mess:
        s=(ord(i)-32+key)%94
        cipher+=chr(s+32)
    return cipher
def dec(mess, key):
    cipher= "" 
    for i in mess:
        s=(ord(i)-32-key)%94
        cipher+=chr(s+32)
    return cipher
print(dec(enc(s,k),k))
print(enc(enc(s,k), -k))
```
