```python
from time import time
n=100000
l1=[ i for i in range(n) ]
```

```python
l2=np.arange(n)
```

```python
l=[]
start=time()
for i in l1:
    l.append(i**2)
end=time()
print(end-start)
```

```python
start=time()
l=[ i**2 for i in l1 ]
end=time()
print(end-start)
```

```python
start=time()
l=list(map(lambda x:x**2, l1))
end=time()
print(end-start)
```

```python
start=time()
l=l2**2
end=time()
print(end-start)
#lista z numpy szybciej niz standardowa z pythona
```
