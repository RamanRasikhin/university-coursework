```python
def fib(n):
    l=[0, 1]
    if n<3:
        return l[:n]
    for i in range(n-2):
        l.append(l[-2]+l[-1])
    return l
print(fib(10))
```

```python
for i in fib(100):
    print(i, end=" ")
```

```python
def fibn1(n):
    a,b = 0, 1
    for _ in range(n):
        a, b= b, a+b
    return a
```

```python
print(fibn1(10))
```

```python
# generator
def fibn(n):
    a,b = 0, 1
    for _ in range(n):
        yield a
        a, b= b, a+b
        
        
for i in fibn(20):
    print(i, end=" ")
```

```python
fib100=fibn(100)
```

```python
next(fib100)
```

```python
fib100.__next__()
```

```python
# iterator
class zakres():
    def __init__(self, n):
        self.n=n
        self.i=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.i==self.n:
            raise StopIteration
        
        self.i+=1
        return self.i    
```

```python
for i in zakres(5):
    print(i, end=" ")
```

```python
class fibc():
    def __init__(self, n):
        self.n=n
        self.a=0
        self.b=1
        self.i=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.i==self.n:
            raise StopIteration
        if self.i==0: 
            self.i+=1 
            return 0
        self.a, self.b= self.b, self.a+self.b
        self.i+=1
        return self.a
for i in fibc(5):
    print(i, end=" ")
```

```python
# nieskonczona petla, ktora liczy czas od ostatniego dzialania iteratora
class timea():
    def __init__(self):
#        self.n=n
        self.a=time()
#        self.i=0
    def __iter__(self):
        return self
    def __next__(self):
#        if self.i==self.n:
#            raise StopIteration
        tt=time()
        self.a, tt= tt, self.a
#        self.i+=1
        return self.a-tt
b=timea()
```

```python
next(b)
```
