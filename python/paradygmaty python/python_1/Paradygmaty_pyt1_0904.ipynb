```python
# Zadanie 1

import math
def kwadr(a, b, c):
    delta=b**2-4*a*c
    if delta<0:
        return 0
    if delta>0:
        x1=-b+math.sqrt(delta)/(2*a)
        x2=-b-math.sqrt(delta)/(2*a)
        return (x1,x2)
    else:
        x=-b/(2*a)
        return x
a=float(input())
b=float(input())
c=float(input())
kwadr(a, b, c)
```

```python
# Zadanie 2
def silnia(a):
    wynik=1
    for el in range(a+1):
        if el>=1:
            wynik*=el  
    return wynik

a=4    
silnia(a)
```

```python
# Zadanie 3(z/d) ( nie dorobione)
student_oceny = {
    "Matematyka": [3, 5, 4, 2, 3],
    "Fizyka": [3, 2, 3, 5, 6, 4],
    "Wychowanie fizyczne": [4, 3, 5, 5, 5],
    "Polski jezyk": [3, 4, 5, 3, 4]
}

def srednia(lista_ocen):
    calosc = 0
    liczba_elementow = 0
    for element in lista_ocen:
        calosc += element
        liczba_elementow += 1
    if liczba_elementow == 0:
        return 0  # Zabezpieczenie przed dzieleniem przez zero
    wynik = calosc / liczba_elementow
    return wynik

while True:
    print("\nMenu:")
    print("1. Dodaj ocenę")
    print("2. Wyświetl oceny z przedmiotu")
    print("3. Oblicz średnią z przedmiotu")
    print("4. Wyjdź")

    wybor = input("Wybierz opcję (1-4): ")

    if wybor == '1':
        przedmiot = input("Podaj nazwę przedmiotu: ")
        ocena = int(input("Podaj ocenę (1-6): "))
        if przedmiot not in student_oceny:
            student_oceny[przedmiot] = []
        student_oceny[przedmiot].append(ocena)
        print(f"Dodano ocenę {ocena} z przedmiotu {przedmiot}")

    elif wybor == '2':
        przedmiot = input("Podaj nazwę przedmiotu: ")
        oceny = student_oceny.get(przedmiot)
        if oceny:
            print(f"Oceny z {przedmiot}: {', '.join(map(str, oceny))}")
        else:
            print("Brak ocen dla tego przedmiotu")

    elif wybor == '3':
        przedmiot = input("Podaj nazwę przedmiotu: ")
        oceny = student_oceny.get(przedmiot)
        if oceny:
            sr = srednia(oceny)
            print(f"Średnia z {przedmiot}: {sr:.2f}")
        else:
            print("Brak ocen dla tego przedmiotu")

    elif wybor == '4':
        print("Koniec programu")
        break

    else:
        print("Nieprawidłowa opcja. Wybierz 1-4")
```

```python
# zadanie 6
arr = [
    [66, 97, 114, 100, 4, 2, 110, 11, 1, 6, 20],
    [99, 3, 10, 122, 76, 101, 111, 3, 32, 100, 0],
    [6, 22, 1, 111, 32, 10, 110, 7, 97, 97, 67],
    [60, 97, 116, 32, 100, 23, 97, 114, 100, 32, 34],
    [2, 106, 15, 6, 111, 56, 80, 20, 10, 86, 10],
    [20, 110, 121, 32, 107, 55, 50, 99, 110, 105, 8],
    [12, 9, 22, 102, 66, 100, 12, 105, 50, 76, 110],
    [42, 81, 123, 92, 26, 98, 20, 1, 20, 11, 10],
]

encoding = "rrrdddllddrrruuuurrddrruurddddlld"

row, col = 0, 0
codes = [arr[row][col]]

for move in encoding:
    if move == 'r':
        col += 1
    elif move == 'l':
        col -= 1
    elif move == 'd':
        row += 1
    elif move == 'u':
        row -= 1
    codes.append(arr[row][col])

result = ''.join(chr(c) for c in codes)
print(result)
```
