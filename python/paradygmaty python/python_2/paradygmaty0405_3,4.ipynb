```python
# zadanie 3
class Osoba:
    def __init__(self, imie, nazwisko):
        self._imie = imie
        self._nazwisko = nazwisko

    def imie(self):
        print(self._imie)

    def nazwisko(self):
        print(self._nazwisko)

    def __str__(self):
        return f"{self._imie} {self._nazwisko}"
```

```python
# zadanie 4
class FamilyTree:
    def __init__(self):
        self.family = {}

    def add_son_father(self, son, father):
        self.family[son] = father

    def find_grandfather(self, grandson):
        father = self.family.get(grandson)
        if father is None:
            return None
        return self.family.get(father)
```

```python
# Utworzenie drzewa genealogicznego
tree = FamilyTree()
tree.add_son_father("Bob", "Mike")
tree.add_son_father("Mike", "John")

# Znajdowanie dziadka
print(tree.find_grandfather("Bob"))  # Wynik: John
print(tree.find_grandfather("Mike")) # Wynik: None (dziadek nieznany)
```
