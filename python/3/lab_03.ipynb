```python
predict = [ "".join("*" if int(s) else "_" for s in bin(i)[2::].zfill(3)) for i in range(7, -1, -1) ]
prerule = [ "*" if int(i) else "_" for i in bin(rule)[2::].zfill(8) ]
automat=dict([i for i in zip(predict, prerule)])
```
