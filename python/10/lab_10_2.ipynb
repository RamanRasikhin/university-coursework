```python
s="Francja"
def stolica2(s):
    f=get('https://pl.wikipedia.org/wiki/' + s, stream=True)
    b, c, d='', '', ''
    for i in f.iter_lines():
        a, b, c, d=(b, c, d, i.decode('UTF-8'))
        if 'Stolica</a>' in a:
            f.close()
            return d.split('">', 1)[1].split('<', 1)[0]
    
panstwa=["Polska", "USA", "Litwa", "Francja", "Serbia", "Kanada", "Czad", "Niemcy", "Belgia", "Szwecja"]
start=time()
for i in panstwa:
    print(stolica2(i))
end=time()
print("Czas wywolania: " + str(end-start))
```

```python
from threading import Thread
```

```python
s=["Polska"]
start=time()
t1=Thread(target=stolica2, args=s)
t1.start()
end=time()
print("Czas wywolania: " + str(end-start))
```

```python
start=time()
for i in panstwa:
    text = [i]
    t1=Thread(target=stolica2, args=text)
    t1.start()   
t1.join()
end=time()
print("Czas wywolania: " + str(end-start))  
```

```python
watki = [Thread(target=stolica2, args=(p,)) for p in panstwa]
start=time()
for w in watki:
    w.start()
for w in watki:
    w.join()    
end=time()
print("Czas wywolania: " + str(end-start))    
```
