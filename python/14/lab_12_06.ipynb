```python
def fib(n):
    a, b=0, 1
    for _ in range(n):
        yield a
        a, b=b, a+b
```

```python
from contextlib import contextmanager
```

```python
@contextmanager 
def tymczasowa(napis):
    print(f"Wejscie {napis}")
    yield
    print(f"Wyjscie {napis}")
```

```python
with tymczasowa("Sekcja krytyczna"):
    print("Dziala w srodku")
```

```python
def srednia():
    suma=0
    licznik=0
    while True:
        x=yield None if licznik==0 else suma/licznik
        licznik+=1
        suma+=x
```

```python
sr=srednia()
```

```python
sr.send(None)
```
