```python
import time
import itertools
import math
from collections import deque
from contextlib import contextmanager
```

```python
# Zadanie 1:
def abc(n):
    for i in range(1, n+1):
        if i % 15 == 0:
            yield "AB"
        elif i % 3 == 0:
            yield "A"
        elif i % 5 == 0:
            yield "B"
        else:
            yield i
```

```python
for x in abc(20):
    print(x)
```

```python
# Zadanie 2:
def fib_generator():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
```

```python
from itertools import islice
fib_gen = fib_generator()
fib_100_120 = list(islice(fib_gen, 100, 121))
```

```python
# Zadanie 3:
def hamiltonian_cycle_lengths(points_iter):
    points = list(points_iter)
    if not points:
        return
    start = points[0]
    rest = points[1:]
    for perm in itertools.permutations(rest):
        cycle = (start,) + perm + (start,)
        length = 0.0
        for p, q in zip(cycle, cycle[1:]):
            length += math.hypot(q[0] - p[0], q[1] - p[1])
        yield length
```

```python
points = [(0,0), (1,0), (1,1), (0,1)]
for L in hamiltonian_cycle_lengths(iter(points)):
    print(L)
```

```python
# Zadanie 4:
@contextmanager
def czasomierz():
    start = time.time()
    try:
        yield
    finally:
        end = time.time()
        print(f"Czas wykonania: {end - start:.6f} s")
```

```python
def do_nothing():
    pass

with czasomierz():
    for i in range(10000):
        for j in range(10000):
            do_nothing()
```

```python
# Zadanie 5:

def srednia_okna(n):
    buf = deque(maxlen=n)
    suma = 0.0
    avg = None
    try:
        while True:
            x = yield avg
            if len(buf) == n:
                suma -= buf[0]
            buf.append(x)
            suma += x
            avg = suma / len(buf)
    except GeneratorExit:
        pass
```

```python
coro = srednia_okna(3)
next(coro) 
print(coro.send(10))
print(coro.send(20))
print(coro.send(30))
print(coro.send(40))
print(coro.send(50))
print(coro.send(1))
print(coro.send(9))
```
