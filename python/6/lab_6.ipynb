Zbior Manderbrota (pierwsza wersja)

```python
import time 
x=np.linspace(-2, 1, 300)
y=np.linspace(-1, 1, 200)
X, Y= np.meshgrid(x,y)
Z=X+Y*1j
s=Z.shape
def f(p):
    z=p
    for i in range(255):
        if abs(z)>2:
            return i
        z=z**2+p
    return 255
start=time.time()
mandelbrot=np.array([[ f(Z[i, j]) for j in range(s[1])]  for i in range(s[0])])
end=time.time()
print(end-start)
```

```python
# Z[0,0] # liczba zespolona z X, Y (liczba p)
```

```python
fig = plt.figure(figsize=(20, 20))
plt.imshow(mandelbrot, cmap="binary")
```

Zbior Manderbrota(druga wersja(lepsza))

```python
x=np.linspace(-2, 1, 300)
y=np.linspace(-1, 1, 200)
X, Y= np.meshgrid(x,y)
Z=X+Y*1j
s=Z.shape
mandelbrot=255*np.ones(s)
P=Z.copy()
start=time.time()
for i in range(255):
    mandelbrot[np.logical_and(abs(Z)>2, mandelbrot==255)] = 0
    Z=Z**2+P
end=time.time()
print(end-start)
fig=plt.figure(figsize=(20, 20))
plt.imshow(mandelbrot, cmap="binary")
```
